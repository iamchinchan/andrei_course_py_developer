```python
x1 = "'a'-'b'"
print(r"'a'-'b'")
print(repr(x1))
print(x1)
print("'a'-'b'")
print("'abc\"'")
print(r"'abc\"'")
print(x1)  # → 'a'-'b'      (same as print(r"'a'-'b'"))
print(len(x1))  # → 7            (no hidden extra characters)
print(x1 == r"'a'-'b'")  # → True   (genuinely identical content)
```
