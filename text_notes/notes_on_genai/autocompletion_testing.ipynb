```python
# factorial
def fact(n):
    if n == 0:
        return 1
    else:
        return n * fact(n - 1)


print(fact(5))
```

```python
print(fact(5))
print(fact(120))
```

```python
def fact_of_fact(n):
    if n == 0:
        return 1
    else:
        return fact(fact(n))


print(fact_of_fact(3))
```
