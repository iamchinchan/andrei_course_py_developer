```python
def convert_value(val):
    # If the value is a string, process it
    if isinstance(val, str):
        # Remove the currency symbol and whitespace
        val = val.replace("€", "").strip()
        # Handle Thousands
        if "K" in val:
            return float(val.replace("K", "")) * 1000
        # Handle Millions
        elif "M" in val:
            return float(val.replace("M", "")) * 1000000

    # If it's already a number, or fails the string check, try converting directly
    try:
        return float(val)
    except:
        return 0.0
```

```python
good_players_df["Value"] = good_players_df["Value"].apply(convert_value)
good_players_df["Wage"] = good_players_df["Wage"].apply(convert_value)
```

```python
good_players_df["difference"] = good_players_df["Value"] - good_players_df["Wage"]
good_players_df.sort_values(by="difference", ascending=False, inplace=True)
print(good_players_df)
```

```python
import seaborn as sns

sns.set_theme()
graph = sns.scatterplot(x="Wage", y="Value", data=good_players_df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set()
# Create the plot
graph = sns.scatterplot(x="Wage", y="Value", data=good_players_df)

# Adjust axes to log scale
graph.set(xscale="log", yscale="log")

#
# Show the plot
plt.show()
```

```python
from bokeh.plotting import figure, show
from bokeh.models import HoverTool
from bokeh.models import ColumnDataSource

# Create a ColumnDataSource for better integration
source = ColumnDataSource(good_players_df)

p = figure(
    title="Soccer 2019",
    x_axis_label="Wage",
    y_axis_label="Value",
    height=400,
    width=700,
    tools="pan,wheel_zoom,reset",
)

# Plot the circles using the data source
p.circle(x="Wage", y="Value", source=source, size=8, color="navy", alpha=0.5)

# Adding HoverTool to show details
hover = HoverTool(tooltips=[("Name", "@Name"), ("Wage", "@Wage"), ("Value", "@Value")])
p.add_tools(hover)

show(p)
```
